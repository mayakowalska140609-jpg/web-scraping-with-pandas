# Organise all the data in a dictionary

```python
# create a dictionary with the original name of league as key
dict_countries = {'Spanish La Liga':'SP1', 'Spanish Segunda Division':'SP2',
                  'German Bundesliga 1':'D1',
                  'English Premier League':'E0', 'English League 1':'E1', 'English League 2':'E2'}
```

```python
# get a dictionary element
dict_countries['English Premier League']
```

```python
for league in dict_countries:
    print(dict_countries[league])
```

```python
dict_historical_data = {}

# creating list of leagues
# leagues = ['E0', 'E2', 'E3']

# looping through key elements
for league in dict_countries:
    frames = []
    for season in range(19, 25):
        df = pd.read_csv(root + str(season) + str(season+1) + '/' + dict_countries[league] + '.csv')
        df.insert(1, 'season', season)
        frames.append(df)
        df_concat = pd.concat(frames)
        dict_historical_data[league] = df_concat
```
